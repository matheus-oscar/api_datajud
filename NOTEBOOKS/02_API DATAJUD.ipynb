```python
import os
import json
import logging
import pandas as pd
import requests
from datetime import datetime
from time import sleep, perf_counter
from concurrent.futures import ThreadPoolExecutor, as_completed

# Configurar logging
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")

API_URL = "https://api-publica.datajud.cnj.jus.br/api_publica_tjsp/_search"
API_HEADERS = {
    'Authorization': 'ApiKey cDZHYzlZa0JadVREZDJCendQbXY6SkJlTzNjLV9TRENyQk1RdnFKZGRQdw==',
    'Content-Type': 'application/json'
}

BATCH_SIZE = 100
N_THREADS = 25
MAX_RETRIES = 3

PASTA_INPUT = "/lakehouse/default/Files/DATAJUD/lista-processos"
PASTA_LOTES = "/lakehouse/default/Files/DATAJUD/resultado-api/lotes"
PASTA_LOGS = "/lakehouse/default/Files/DATAJUD/resultado-api/logs"

os.makedirs(PASTA_LOTES, exist_ok=True)
os.makedirs(PASTA_LOGS, exist_ok=True)
```

```python
def consultar_processo(numero, url, headers, max_retries=MAX_RETRIES, delay=1.0):
    payload = json.dumps({"query": {"match": {"numeroProcesso": numero}}})
    for tentativa in range(1, max_retries + 1):
        try:
            response = requests.post(url, headers=headers, data=payload, timeout=10)
            response.raise_for_status()
            data = response.json()
            hits = data.get("hits", {}).get("hits", [])
            registros = [dict(hit["_source"], _id=hit.get("_id")) for hit in hits]
            sleep(delay)
            return (numero, registros, None)
        except Exception as e:
            if tentativa < max_retries:
                sleep(delay)
            else:
                return (numero, None, f"Tentativas esgotadas ({tentativa}): {e}")

def processar_em_lotes(lista_processos, api_url, api_headers):
    total = len(lista_processos)
    logging.info(f"Total de processos: {total}")

    # Identificar lotes já processados
    lotes_salvos = {
        int(f.split("_")[1].split(".")[0])
        for f in os.listdir(PASTA_LOTES)
        if f.startswith("lote_") and f.endswith(".parquet")
    }

    # Identificar lotes vazios registrados
    lotes_vazios = set()
    caminho_vazios = os.path.join(PASTA_LOGS, "lotes_vazios.jsonl")
    if os.path.exists(caminho_vazios):
        with open(caminho_vazios, "r", encoding="utf-8") as f:
            lotes_vazios = {json.loads(linha)["lote"] for linha in f if linha.strip()}

    # Determinar o próximo lote a ser processado
    ultimo_lote_salvo = max(lotes_salvos) if lotes_salvos else 0
    ultimo_lote_vazio = max(lotes_vazios) if lotes_vazios else 0
    lote_inicial = max(ultimo_lote_salvo, ultimo_lote_vazio) + 1
    logging.info(f"Retomando a partir do lote {lote_inicial}")

    for i in range((lote_inicial - 1) * BATCH_SIZE, total, BATCH_SIZE):
        lote_num = i // BATCH_SIZE + 1

        if lote_num in lotes_salvos or lote_num in lotes_vazios:
            logging.info(f"Lote {lote_num} já processado ou registrado como vazio. Pulando.")
            continue

        batch = lista_processos[i:i + BATCH_SIZE]
        logging.info(f"Processando lote {lote_num} ({len(batch)} processos)")
        t0 = perf_counter()

        resultados, erros, sem_dados = [], [], []

        with ThreadPoolExecutor(max_workers=N_THREADS) as executor:
            futures = {
                executor.submit(consultar_processo, numero, api_url, api_headers): numero
                for numero in batch
            }

            for future in as_completed(futures):
                numero = futures[future]
                try:
                    _, registros, erro = future.result()
                    if erro:
                        erros.append({"processo": numero, "erro": erro})
                    elif registros:
                        resultados.extend(registros)
                    else:
                        sem_dados.append(numero)
                except Exception as e:
                    erros.append({"processo": numero, "erro": str(e)})

        if resultados:
            df = pd.DataFrame(resultados)
            df["data_download"] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

            colunas_problematicas = ["classe", "assuntos", "orgaoJulgador", "vara", "grau"]
            for coluna in colunas_problematicas:
                if coluna in df.columns:
                    df[coluna] = df[coluna].apply(lambda x: str(x))

            caminho_lote = os.path.join(PASTA_LOTES, f"lote_{lote_num:03}.parquet")
            df.to_parquet(caminho_lote, index=False, compression="zstd")
            logging.info(f"✅ Lote {lote_num} salvo com {len(df)} registros em {perf_counter() - t0:.2f}s")
        else:
            with open(caminho_vazios, "a", encoding="utf-8") as f:
                f.write(json.dumps({
                    "lote": lote_num,
                    "quantidade_processos": len(batch),
                    "timestamp": datetime.now().isoformat()
                }) + "\n")
            logging.info(f"Lote {lote_num} sem registros válidos.")

        if sem_dados:
            with open(os.path.join(PASTA_LOGS, "processos_sem_dados.jsonl"), "a", encoding="utf-8") as f:
                for numero in sem_dados:
                    f.write(json.dumps({
                        "processo": numero,
                        "lote": lote_num,
                        "timestamp": datetime.now().isoformat()
                    }) + "\n")

        if erros:
            with open(os.path.join(PASTA_LOGS, "erros_consulta.jsonl"), "a", encoding="utf-8") as f:
                for erro in erros:
                    f.write(json.dumps({**erro, "lote": lote_num, "timestamp": datetime.now().isoformat()}) + "\n")

```

```python
# Carregar lista de processos
arquivo_input = os.path.join(PASTA_INPUT, "tjsp_processos.parquet")
df_proc = pd.read_parquet(arquivo_input, columns=["Processo"])
lista_processos = df_proc["Processo"].dropna().astype(str).tolist()
```
